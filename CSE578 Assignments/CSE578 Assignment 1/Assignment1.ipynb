**Question 1:** 

What is the most popular attraction to visit in the park?

**Note:** Your output should be the name of the attraction. Please print your output in the cell below.

```python
### TEST FUNCTION: test_question1
# DO NOT REMOVE THE LINE ABOVE
# your code here
sql_query = """
SELECT a.Name, COUNT(*) as Cnt, a.AttractionID, a.Category, a.Type 
FROM checkin c JOIN attraction a ON c.attraction = a.AttractionID 
GROUP BY a.Name, a.Category, a.AttractionID 
ORDER BY Cnt DESC 
LIMIT 10;
"""
c.execute(sql_query) 
most_popular_attraction = c.fetchone() 
most_popular_attraction_name = most_popular_attraction[0]
print(most_popular_attraction_name) 
```

**Question 3:** 

Which Fast Food offering in the park has the fewest visitors?

**Note:** Your output should be the name of the fast food offering. Please print your output in the cell below.

```python
### TEST FUNCTION: test_question3
# DO NOT REMOVE THE LINE ABOVE
# your code here
sql_query = """ 
SELECT a.Name, COUNT(*) as Cnt, a.AttractionID, a.Category, a.Type 
FROM checkin c JOIN attraction a ON c.attraction = a.AttractionID 
WHERE a.Type LIKE '%Fast Food%' 
GROUP BY a.Name, a.Category, a.AttractionID 
ORDER BY Cnt ASC;
"""
c.execute(sql_query) 
least_popular_fast_food_resturant = c.fetchone() 
least_popular_fast_food_resturant_name = least_popular_fast_food_resturant[0]
print(least_popular_fast_food_resturant_name) 
```
